### Qualitative and Quantitavie Data
#### Qualitative Data
Qualitative data describes qualities or characteristics. It is usually non-numerical and represents categories or labels.

Examples:
Color: Red, Blue, Green
Gender: Male, Female
Feedback: Good, Bad, Average
City: Jaipur, Delhi, Mumbai

#### Quantitative Data
Quantitative data represents quantities that can be measured or counted. It is numerical and can be used for mathematical calculations.

Examples:
Age: 21 years
Height: 170 cm
Salary: ₹30,000
Number of students: 50

### Distribution Plots

Used to visualize the distribution of quantitative data. Some of these plots are defined in the following examples - 

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df
```

```python
df["size"].unique()
```

```python
# total_bill, tip are quantitative data
# All others are qualitative data
```

### Distplot or Histplot

```python
sns.displot(df["total_bill"], bins = 30) # By default bins are 20 
```

```python
# KDE (Kernel Density Estimation) is a statistical method used to estimate and visualize 
# the probability density distribution of continuous data using a smooth curve.

# We can see KDE in out graph as follows
sns.displot(df["total_bill"], bins = 30, kde=True)
```

```python
sns.histplot(df["tip"], bins=30, kde=True)
```

#### Joint Plot
Joint Plot is used to visualize the relationship between two variables along with their individual distributions, typically using a scatter plot in the center and histograms along the edges.

```python
sns.jointplot(x='total_bill', y='tip', data=df)
```

```python
# By default, kind is set to 'scatter', we have multiple kinds available - hex, reg etc|
sns.jointplot(x='total_bill', y='tip', data=df, kind="hex")
```

#### Pair Plot
A pair plot is a visualization that shows the relationships between every pair of numerical variables in a dataset using scatter plots and the distribution of individual variables using histograms or KDE plots.

```python
sns.pairplot(df)
```

```python
# We can change hue as per categorical data as well
sns.pairplot(df, hue='sex')
```

#### Rug Plot

```python
sns.rugplot(df['total_bill'])
```
