## Pandas
Pandas is a software library written for the Python programming language for data manipulation and analysis.

### Series
A series is a 1D labeled array, capable of holding any data type. The axis labels are collectively called index.

```python
import pandas as pd
import numpy as np
```

```python
labels = ['a', 'b', 'c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_list) # The labels by default start from 0
```

```python
pd.Series(my_list, index=labels)
```

```python
# Series can be created from series as well
pd.Series(arr)
```

```python
# While creating series from a dictionary, the keys become the label
pd.Series(d)
```

#### Dataframe
A Pandas DataFrame is a two-dimensional table that stores data in rows and columns

##### Creating a Dataframe

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}

df = pd.DataFrame(data)
df
```

```python
# When we don't provide keys, it marks each column with a number
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
df2
```

```python
# We can define the column name as follows
columns = ["Name", "Age", "City", "Salary"]
df2 = pd.DataFrame(data_list, columns=columns)
df2
```

##### Selection of Columns

```python
df2
```

```python
df2["Name"]
```

```python
df2[["Name", "City"]]
```

##### Creating a new Column

```python
# When adding a new column, we need to provide the correct number of data
df2["Designation"] = ["Doctor", "Engineer", "Doctor", "Engineer"]
```

```python
df2
```

##### Removing a Row/ Column

```python
df2.drop("Designation", axis=1) # By default the axis is zero, so we need to specify it.
```

```python
# Drop (unless specified) does not change the actual dataframe
df2
```

```python
# If we want to do so, we can use inplace=True
df2.drop("Designation", axis=1, inplace=True)
df2
```
