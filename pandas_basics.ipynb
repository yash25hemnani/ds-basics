## Pandas
Pandas is a software library written for the Python programming language for data manipulation and analysis.

### Series
A series is a 1D labeled array, capable of holding any data type. The axis labels are collectively called index.

```python
import pandas as pd
import numpy as np
```

```python
labels = ['a', 'b', 'c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_list) # The labels by default start from 0
```

```python
pd.Series(my_list, index=labels)
```

```python
# Series can be created from series as well
pd.Series(arr)
```

```python
# While creating series from a dictionary, the keys become the label
pd.Series(d)
```

#### Handling Missing Data

##### Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Check if some values are none
df.isna()
```

```python
df.isna().sum()
```

```python
# Tells you which columns have null values
df.isna().any()
```

##### Remove Missing Numbers

```python
df
```

```python
# Using dropna() we can remove rows whcih have null values
df.dropna()
```

```python
# using thresh in dropna()
# thresh specifies how many non-null values must exist and then remove
df.dropna(thresh=3)
```

```python
# Using fillna() allows you to fill values at null locations
df.fillna(0)
```

```python
# We can define what we want in place of null in a particular column
values={'A':0, 'B': 100, 'C':300, 'D':400}
df.fillna(value=values)
```

```python
# But to persist these changes in the original df, we have to use inplace=True
df.fillna(0, inplace=True)
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
# We can replace null values with mean
df.fillna(df.mean())
```

#### Merging, Joining and Concatenation

```python
# DataFrame 1: Employee Information
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
# Only the key that is common among both will be used to merge (only ids that exist in 
# both will be visible)
pd.merge(employees, salaries)
```

```python
# We can specifically define which column to merge on as well
pd.merge(employees, salaries, on='employee_id')
```

```python
# we can also define 'how' which by default is inner
pd.merge(employees, salaries, on='employee_id', how='inner')
```

```python
# we can use outer join to get all the values - with null for those which do not exist
pd.merge(employees, salaries, on='employee_id', how='outer')
```

```python
pd.merge(employees, salaries, on='employee_id', how='left')
```

```python
pd.merge(employees, salaries, on='employee_id', how='right')
```

##### Concatenation of 2 Dataframes

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
# If no axis is defined, concatenation happens column wise
pd.concat([df1,df2])
```

```python
pd.concat([df1, df2], axis=1)
```

##### Joining 2 Dataframes

```python
# First Dataframe
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
# By default, it is inner join
df1.join(df2)
```

```python
# We can specify the type of join as follows
df1.join(df2, how='outer')
```

#### Group By and Aggregation

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Group by Category and Calculate the sum of sales
gb = df.groupby('Category') # This is an object
# We can view it by looping
for i in gb:
    print(i)
```

```python
df.groupby('Category')['Sales'].sum()
```

```python
# Group by store and calculate sum of sales
df.groupby('Store')['Sales'].sum()
```

```python
# Group by multiple columns
df.groupby(['Category','Store'])['Sales'].sum()
```

##### Aggregation 
An aggregate in Pandas is an operation that reduces multiple values down to a single summary statistic, such as a sum, mean, minimum, maximum, or count.

```python
df['Sales'].mean()
```

```python
df['Sales'].median()
```

```python
df['Sales'].mode()
```

```python
# We can get multiple values at once as well
df['Sales'].agg(['mean', 'median', 'min','max'])
```

#### Pivot Tables

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
# pd.pivot_table(
#     df,
#     values=X,
#     index=Y,
#     columns=Z,
#     aggfunc=W
# )
# Read it as:
# "Take column X, group it by Y, split it by Z, and perform calculation W."
pd.pivot_table(df, values="Sales", index="Region", columns="Product")
```

```python
# We can define the aggregation function as well, which is mean by default
pd.pivot_table(df, values="Sales", index="Region", columns="Product", aggfunc="median")
```

```python
pd.pivot_table(df, values=["Sales", "Units"], index="Region", columns="Product", aggfunc="median")
```

```python
# Always take numeric values in pivot table since you are performing aggregations
```

#### Cross Tabs
Crosstab is very similar to a pivot table, but it's mainly used to count how often combinations of categories occur.

```python
pd.crosstab(df["Region"], df["Product"])
```
