#### Deep vs Shallow Copy

```python
# A shallow copy creates a new outer object, but nested objects are still shared.
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

shallow[0][0] = 99

print(original)  # [[99, 2], [3, 4]]
print(shallow)   # [[99, 2], [3, 4]]
```

```python
# A deep copy creates a new outer object and recursively copies nested objects.
import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[0][0] = 99

print(original)  # [[1, 2], [3, 4]]
print(deep)      # [[99, 2], [3, 4]]
```

```python
# Shallow Copy in Numpy
import numpy as np

a = np.array([1, 2, 3, 4])
b = a.view()

b[0] = 99

print(a)  # [99  2  3  4]
print(b)  # [99  2  3  4]
```

```python
# Deep Copy in Numpy
import numpy as np

a = np.array([1, 2, 3, 4])
b = a.copy()

b[0] = 99

print(a)  # [1 2 3 4]
print(b)  # [99  2  3  4]
```
