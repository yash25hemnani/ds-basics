#### Task 1 - Extract Number of Episodes

```python
# We see each title has information about episodes, we will extract it as follows
def extract_episodes(txt):
    bracket_found = False
    eps = ""
    for i in txt:
        if i == ')':
            bracket_found = False
            break;
        if i == '(':
            bracket_found = True
            continue
        if bracket_found == True:
            eps += i
    return eps
```

```python
df["Episodes"] = df["Title"].apply(extract_episodes)
```

```python
df
```

```python
df["Episodes"] = df["Episodes"].str.replace(" eps","")
```

```python
df.info()
```

```python
# We need to change episodes to integer, so we can perform operations on it
```

```python
df["Episodes"] = df["Episodes"].astype(int)
```

#### Task 2 - Extract Run Duration

```python
df.loc[10]["Title"]
```

```python
def extract_duration(txt):
    bracket_found = False
    duration = ""
    str_len = 19
    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(str_len):
                duration += txt[i + j + 1]

    return duration
```

```python
df["Run Duration"] = df["Title"].apply(extract_duration)
```

```python
df.head()
```

#### Task 3 - Get Total Months 

```python
from dateutil.relativedelta import relativedelta
import pandas as pd
```

```python
def get_months(txt):
    start, end = txt.split(" - ")

    s_dt = pd.to_datetime(start, format="%b %Y")
    e_dt = pd.to_datetime(end, format="%b %Y")

    delta = relativedelta(e_dt, s_dt)

    return delta.years * 12 + delta.months

df["Months"] = df["Run Duration"].apply(get_months)
```
