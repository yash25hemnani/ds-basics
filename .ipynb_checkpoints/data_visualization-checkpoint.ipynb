#### Subplots in Matplotlib


```python
# Takes in rows and columns - divides the canvas into rows x columns sections
# each of them can have their own graphs. The fucntion takes - row, col, index of graph
plt.figure(figsize=(15, 7), dpi=120)
plt.subplot(2,3,1)
plt.plot(x_values**2, x_values**2)
plt.subplot(2,3,2)
plt.plot(x_values ** 3, x_values)
plt.subplot(2,3,3)
plt.plot(x_values, y_values)
plt.subplot(2,3,4)
plt.plot(y_values, x_values)
```

#### Using figure()

```python
# Create a Figure object, which represents the entire canvas on which plots are drawn
fig = plt.figure()

# Add an Axes: [left, bottom, width, height]
# Values are fractions of the Figure's dimensions
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x_values, y_values)

# Add another Axes on the same Figure
# Same starting position, smaller size → overlapping plots
axis2 = fig.add_axes([0.3, 0.6, 0.3, 0.3])
axis2.plot(x_values, y_values)
```

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).

```python
fig = plt.figure(figsize=(6,4), dpi=200)
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x_values, y_values)
```
